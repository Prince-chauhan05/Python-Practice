# nested condition 

# A nested condition is a condition inside another condition. It means when one if statement is placed inside another if statement. The inner condition is checked only if the outer condition is true.

```python
n= int(input('enter n'))
if n%5==0:
    if n%7==0:
        print(n,'is divisble by 7')
    else: 
        print(n,'is divisble by 5 only')
elif n%7==0:
    print(n,'is divisble by 7 only')
else:
    print(n,'is not divisble by any of them')
    
```
