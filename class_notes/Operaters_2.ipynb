```python
# 5. Identity operator - are responsible for comparing the address of variable 
# is
```

```python
x= 10
y= 10.0
print(x==y)
print(x is y) # False because id's not match
print(id(x),'\n',id(y))  # or
print(f'{id(x)}\n{id(y)}')  # where f"" is formatted string and curly bracketes{} ka use value access krne ke liye 
print(x is not y)
```
