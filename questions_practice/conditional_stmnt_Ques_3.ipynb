```python
# 1. Calculator Extension (Simple Calculator)

# User se 2 numbers aur ek operator (/ ya %) input lein.

# Agar operator / hai, toh division print karein.

# Agar operator % hai, toh modulus print karein.

# Agar koi aur character ho, toh 'Invalid operator' print karein.

a= float(input('enter a'))
b= float(input('enter b'))
c= input('enter c')
if c=='/':
    print(a/b)
elif c=='%':
    print(a%b)
else:
    print('invalid operator')
```

```python
# 3. Electricity Bill Calculator (Multiple Slabs)
# User se electricity units input lein:

# Units 100 se kam ya barabar (<= 100): Free (bill = 0).

# Units 101 se 200 ke beech (101 to 200): ₹5 per unit.

# Units 200 se upar (> 200): ₹10 per unit.

# Total bill calculate karke print karein.

u= int(input('enter unit'))
if u<=100:
    print(u*0)
elif u>100 and u<=200:
    print((u-100)*5)
else:
    print((u-200)*10+(100*5))
        
```

```python
# # # Divisibility & Multiples (3 and 5 Check)
# # User se ek integer input lein.

# # Check karein:

# # Agar number 3 aur 5 dono se divide hota hai, toh print karein:'FizzBuzz'
# Agar sirf 3 se divide hota hai, toh print karein: 'Fizz'

# Agar sirf 5 se divide hota hai, toh print karein: 'Buzz'

# Agar kisi se nahi hota, toh print karein:'Not divisible by 3 or 5'

n= int(input('enter n'))
if n%3==0 and n%5==0:
    print('FizzBuzz')
elif n%3==0:
    print('Fizz')
elif n%5==0:
    print('Buzz')
else:
    print('Not divisible by 3 or 5')
```

```python
#  1. E-Commerce Order Discount & Delivery Fee Engine
#  Ek shopping app ka checkout logic design karein:
# Input: Total cart amount (cart_total) aur user ka status (is_prime = True/False).

# # # Rules:
# # Agar user Prime member hai:
# Cart total $\ge 500$ hai toh 20% discount milega aur Delivery Free hogi.
# # Cart total $< 500$ hai toh 10% discount milega aur Delivery Free hogi.
# Agar user Prime member nahi hai (Non-Prime):
# Cart total $\ge 1000$ hai toh 10% discount milega aur Delivery Free hogi.
# Cart total $< 1000$ hai toh koi discount nahi milega aur ₹50 delivery charge lagega.
# Output: Final payable amount print karein.
```

```python
cart_total= float(input('enter amount'))
is_prime= str(input('are you prime member? (yes/no)')).lower()

if is_prime=='yes':
    if cart_total>=500:
        amount=cart_total-((cart_total*20)/100)+0
    
    else:
        amount= cart_total-((cart_total*10)/100)+0
    print('final payable amount',amount)
elif is_prime=='no':
    if cart_total>=1000:
        amount= cart_total-((cart_total*10)/100) +0
    else:
        amount= cart_total-((cart_total*0)/100)+50
    print('final payable amount',amount)
else:
    print('invalid input! Please enter only "yes" or "no"')
    
```

```python
# # Employee Performance & Bonus Eligibility (HR Analytics)
# # Ek company ke HR system ke liye bonus calculator banayein:
# # Input: Employee ki current salary, unka experience_years, aur rating (scale 1 to 5).

# #  Rules:
# # Agar rating == 5: 20% bonus.

# # Agar rating == 4:

# # Agar experience_years >= 3 hai toh 15% bonus, warna 10% bonus.

# # Agar rating == 3: 5% bonus.

# # Agar rating < 3: No bonus (0%).
# Output: Calculated bonus amount aur total compensation (salary + bonus) print karein.
```

```python
s= float(input('enter s'))
experience_years= int(input('enter experience_years ?'))
rating= int(input('enter rating ? (1 to 5)'))
if rating==5:
    print(s+(s*.20))
elif rating==4:
    if experience_years>=3:
        print(s+(s*.15))
    else:
        print(s+(s*.10))
elif rating==3:
    print(s+(s*.05))
elif rating<3 and rating>=1:
    print(s)
else:
    
    print('invalid inputs')

     
```

```python
# # # Banking ATM Cash Withdrawal Validation
# # # ATM machine ka background validation logic likhein:
# Input: account_balance, withdrawal_amount, aur daily_limit_remaining.

# Rules:
# # Check 1: Withdrawal amount 100 ka multiple hona chahiye (withdrawal_amount % 100 == 0). Agar nahi hai, toh print: 'Please enter amount in multiples of 100'.

# # Check 2: Withdrawal amount balance se bada nahi hona chahiye (withdrawal_amount <= account_balance). Agar bada hai, toh print: 'Insufficient Balance'.

# # Check 3: Withdrawal amount daily limit se kam ya barabar hona chahiye (withdrawal_amount <= daily_limit_remaining). Agar bada hai, toh print: 'Daily transaction limit exceeded'.

# # Agar teeno conditions pass ho jayein: Dedicated amount deduct karke remaining balance print karein aur message de: 'Transaction Successful'.
```
