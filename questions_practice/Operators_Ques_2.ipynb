```python
# Shopping Mall Bill & Discount Check (Arithmetic + Comparison):
# Ek customer ne 3 items kharide hain jin kile price item1 = 450, item2 = 750, aur item3 = 300 hain.

# Total bill calculate karein.

# Check karein ki kya total bill 1200 se bada hai? Output True ya False aana chahiye.

i1 = 450
i2= 750
i3= 300
i = i1+i2+i3
print(i)
i >1200 
print(i>1200)
```

```python
# Attendance Eligibility System (Logical Operator):
# Ek student ki Total Classes = 100 hain aur Attended Classes = 72 hain.

# Attendance percentage calculate karein (Attended / Total) * 100.

# Exam mein baithne ke liye Attendance >= 75% honi chahiye. Student eligible hai ya nahi, ye True/False mein check karein.
t_class= 100
a_class= 72
percentage = (a_class/t_class)*100
print(percentage)
print(percentage >=75)
```

```python
# Username & Email Verification (Membership Operator):

# Check karein ki email 'prince123@gmail.com' mein '@' aur '.com' dono present hain ya nahi.

# Check karein ki user ke username 'Prince_2026' mein space ' ' hai ya nahi (space nahi hona chahiye, output True aana chahiye agar space nahi hai).

email= 'prince123@gmail.com'
user_name = 'Prince_2026'
print('@' in email)
print('.com' in email)
print(' ' not in user_name)
```

```python
# Compound Interest / Wallet Balance (Compound Assignment):
# Aapke digital wallet mein initial balance wallet = 1000 rupees hai.

# 500 add karein using compound addition (+=).

# 200 spend karein using compound subtraction (-=).

# Remaining balance ko 3 se multiply karein using *=. Final balance print karein.

wallet = 1000
wallet+=500
wallet-=200
wallet*=3
print(wallet)
```

```python
# Odd/Even & Positive Number Challenge (Arithmetic + Comparison + Logical):
# User se ek number input lein aur bina if condition ke check karein:

# Number positive (greater than 0) hona chahiye AND number even hona chahiye.

# Code True return kare agar dono conditions satisfied hain, warna False.

n= int(input('enter n'))
a= n>0 and n%2==0
print(a)
```
