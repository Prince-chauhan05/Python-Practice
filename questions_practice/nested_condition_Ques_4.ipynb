```python
# Practice Question: Custom Food & Beverage Ordering System
# Build a nested if-else program for a cafe menu with the following rules:
# 1. Ask user to place an order: 'coffee', 'sandwich', or 'shake'.
# 2. If order is 'coffee':
#    - Ask for type: 'hot' or 'cold'.
#    - If 'hot', price is 120. If 'cold', price is 150. Otherwise, print 'type not available'.
# 3. If order is 'sandwich':
#    - Ask for bread type: 'brown' or 'white'.
#    - If 'brown', price is 90. If 'white', price is 70. Otherwise, print 'bread not available'.
# 4. If order is 'shake':
#    - Ask for flavor: 'chocolate' or 'strawberry'.
#    - If 'chocolate', price is 160. If 'strawberry', price is 140. Otherwise, print 'flavor not available'.
# 5. If order is anything else, print 'item not in menu'.
```

```python
order= str(input('place your order: Coffee/sandwich/shake')).lower()
if order=='coffee':
    types= str(input('hot / cold')).lower()
    if types=='hot':
        print(types,order,'price is 120')
    elif types== 'cold':
        print(types,order,'price is 150')
    else:
        print(types, 'not available')
elif order=='sandwich':
    bread_type= str(input('brown/white')).lower()
    if bread_type=='brown':
        print(bread_type,order,'price is 90')
    elif bread_type=='white':
        print(bread_type,order,'price is 70')
    else:
        print('bread not available')
elif order == 'shake':
    flavor= str(input('select flavor: chocolate/ strawberry')).strip().lower()
    if flavor=='chocolate':
        print(flavor,order,'price is 160')
    elif flavor=='strawberry':
        print(flavor,order,'price is 140')
    else:
        print(flavor,order,'not available')
else:
    print(order,'item not in a menu')
        
```

```python
# E-Commerce VIP Membership & Discount Engine
# Scenario:
# Ek online shopping portal hai jo customers ki membership type aur shopping amount ke basis par final bill calculate karta hai aur extra perks deta hai.
# Main Input: User se poochho:total_purchase_amount (float)membership_type (string: 'gold', 'silver', ya 'regular') - isko .strip().lower() kar lena.Step 1: Nested Check for Membership & Discounts:Agar membership 'gold' hai:Agar purchase amount $\ge 5000$ hai $\rightarrow$ 30% discount milega aur saath mein print hoga: 'Congratulations! You get a Free Gold Gift Box.'Agar purchase amount $5000$ se kam hai $\rightarrow$ 20% discount milega aur print hoga: 'Gold Member discount applied.'Agar membership 'silver' hai:Agar purchase amount $\ge 3000$ hai $\rightarrow$ 15% discount milega aur print hoga: 'Silver Member discount applied.'Agar purchase amount $3000$ se kam hai $\rightarrow$ 10% discount milega.Agar membership 'regular' hai:Agar purchase amount $\ge 2000$ hai $\rightarrow$ 5% discount milega.Agar purchase amount $2000$ se kam hai $\rightarrow$ Koi discount nahi (0% discount), aur print hoga: 'No discount for regular orders below 2000.'Agar membership galat daali:Print karo: 'Invalid membership type!' aur program ko rok do ya amount change mat karo.Step 2: Final Calculation & Output:Calculated discount ko minus karke final_payable_amount nikalo.Print karo: Original Amount, Discount Amount, aur Final Payable Amount.
```

```python
amount= float(input('total purchase amount'))
membership= str(input('membership type: gold/silver/regular')).strip().lower()
if membership=='gold':
    if amount>=5000:
        discount= amount*.30
        print('congratulations! You get a free gold Gift Box')
    else:
        discount= amount*.20
        print('Gold member discount applied')
elif membership=='silver':
    if amount>=3000:
        discount= amount*.15
        print('silver member discount applied')
    else:
        discount= amount*.10
        
elif membership=='regular':
    if amount>=2000:
        discount=amount*.05
        
    else:
        discount= amount*0
        print('No discount for regular orders below 2000')
else:
    discount= 0
    print('Invalid membership type!')

final_payable_amount= amount-discount
print('original amount',amount)
print('discount amount',discount)
print('final payable amount',final_payable_amount)
    
```

```python
# E-Commerce Smart Cart & Checkout Engine
# Scenario:

# Ek online shopping store ka checkout portal hai jo customer ki shopping cart ka final bill calculate karta hai. Isme product category, delivery speed, coupon code, aur payment method ke hisaab se calculations hoti hain.

# Rules & Instructions (The Blueprint):
# Main Inputs:

# cart_amount (float) 
# → Total items ki price ka amount.

# category (string: 'electronics', 'clothing', 'grocery') 
# → .strip().lower()

# delivery_type (string: 'standard', 'express') 
# → .strip().lower()

# coupon_code (string: 'save10', 'festive20', 'none') 
# → .strip().lower()

# payment_mode (string: 'upi', 'cod') 
# → .strip().lower()

# Step 1: Category-Based Tax / Packaging Fee (Mutually Exclusive - Use if-elif-else):

# Customer ki item category ke hisaab se packaging/handling fee lagegi:

# 'electronics' 
# → Packaging fee = ₹150

# 'clothing' 
# → Packaging fee = ₹50

# 'grocery' 
# → Packaging fee = ₹30

# Galat category daalne par 
# → Fee = 0 aur print karo 'Invalid category!'.

# Step 2: Delivery Charges (Mutually Exclusive - Use if-elif-else):

# Delivery speed ke hisaab se shipping charge add hoga:

# 'standard' delivery 
# → Shipping charge = ₹80

# 'express' delivery 
# → Shipping charge = ₹200

# Galat delivery type daalne par 
# → Shipping = 0 aur print karo 'Invalid delivery type!'.

# Step 3: Discount Coupon (Mutually Exclusive - Use if-elif-else):

# Cart amount par discount coupon apply hoga:

# 'save10' 
# → Cart amount par 10% discount (* 0.10)

# 'festive20' 
# → Cart amount par 20% discount (* 0.20)

# 'none' 
# → Discount = 0

# Galat coupon daalne par 
# → Discount = 0 aur print karo 'Invalid coupon code!'.

# Step 4: Payment Mode Extra Fee or Cashback (Independent / Additive check):

# Payment mode ke hisaab se extra charge ya discount judega:

# Agar payment_mode == 'cod' (Cash on Delivery) hai 
# → ₹40 extra handling fee judegi.

# Agar payment_mode == 'upi' hai 
# → ₹20 ka instant cashback milega (yani final amount mein se ₹20 minus honge).

# Galat payment mode par 
# → Fee/Cashback = 0 aur print karo 'Invalid payment mode!'.

# Step 5: Final Output:

# Print karo: Original Cart Amount, Packaging Fee, Shipping Charge, Discount Amount, Payment Adjustment (Fee/Cashback), aur sabko jod-ghata kar Final Payable Amount.

```
